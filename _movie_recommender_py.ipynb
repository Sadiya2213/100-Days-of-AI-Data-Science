```python


print("🎬 --- Movie Night Compatibility & Recommender --- \n")

my_watchlist = {"Inception", "Interstellar", "Dune", "3 Idiots", "The Dark Knight"}
friend_watchlist = {"Dune", "3 Idiots", "Avengers", "Oppenheimer", "Spider-Man"}


common_movies = my_watchlist.intersection(friend_watchlist)


new_for_me = friend_watchlist.difference(my_watchlist)

print(f"🍿 Movies you BOTH want to watch tonight: {common_movies}")
print(f"✨ New recommendations from your friend's list: {new_for_me}\n")


if len(common_movies) >= 2:
    print("🔥 Match Score: High! You have great taste in common. Pick one and grab popcorn!")
elif len(common_movies) == 1:
    print("👍 Match Score: Decent! At least you agree on one movie.")
else:
    print("😅 Match Score: Low! Time to flip a coin.")
```
